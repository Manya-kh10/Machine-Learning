```python
from sklearn.linear_model import LinearRegression
import numpy as np

lr = LinearRegression()

# df4 = dataframe you are iteratively imputing
# df  = original dataframe (to check original NaNs)

for col in df4.columns:
    print(f"\nImputing column: {col}")

    for row in df4.index:
        if np.isnan(df.loc[row, col]):   # Only impute original NaN positions

            # X: all columns except the one being imputed
            X = df4.drop(columns=[col])
            y = df4[col]

            # Use only rows where y is known
            mask = ~y.isna()
            X_train = X.loc[mask]
            y_train = y.loc[mask]

            # Prepare the prediction input (single row, drop the target col)
            X_pred = X.loc[row].values.reshape(1, -1)

            # Fit & predict
            lr.fit(X_train, y_train)
            predicted_value = lr.predict(X_pred)[0]

            # Update df4 iteratively
            df4.loc[row, col] = predicted_value

            print(f"  filled row {row} with {predicted_value:.4f}")
```
